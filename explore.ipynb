# Explore Parsed Outputs

Minimal notebook for loading parsed outputs and plotting basic statistics.

```python
from collections import Counter
from utils.analysis import load_parsed_records, create_report

records = load_parsed_records("outputs")
print("records:", len(records))
```

```python
def print_basic_stats(records):
    model_counts = Counter(r.get("model_used", "unknown") for r in records)
    concept_counts = Counter(r.get("concept", "unknown") for r in records)
    dimension_counts = Counter((r.get("quality_dimension") or r.get("domain") or "categorical") for r in records)

    print("models:", dict(model_counts))
    print("top concepts:", concept_counts.most_common(10))
    print("top dimensions:", dimension_counts.most_common(10))

print_basic_stats(records)
```
